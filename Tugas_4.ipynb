```python
while True:
    try:
        usia_pengemudi = int(input("Masukkan usia pengemudi: "))
        break
    except ValueError:
        print("Input tidak valid! Harap masukkan angka untuk usia.")

jenis_kendaraan = input("Masukkan jenis kendaraan (mobil/motor): ").lower()

# Memasukkan area penggunaan kendaraan
area_penggunaan = input("Masukkan area penggunaan (perkotaan/pedesaan): ").lower()

# Memasukkan status sistem pengaman
sistem_pengaman = input("Apakah memiliki sistem pengaman? (ya/tidak): ").lower()

# Menentukan risiko berdasarkan usia pengemudi
if usia_pengemudi < 20 or usia_pengemudi > 60:
    risiko_usia = "tinggi"
else:
    risiko_usia = "rendah"

# Menentukan risiko berdasarkan jenis kendaraan
if jenis_kendaraan == "motor":
    risiko_kendaraan = "sedang"
else:
    risiko_kendaraan = "rendah"

# Menentukan risiko awal berdasarkan usia dan jenis kendaraan
if risiko_usia == "tinggi":
    risiko = "tinggi"
elif risiko_kendaraan == "sedang":
    risiko = "sedang"
else:
    risiko = "rendah"

# Meningkatkan risiko jika digunakan di perkotaan
if area_penggunaan == "perkotaan" and risiko != "tinggi":
    if risiko == "rendah":
        risiko = "sedang"
    else:
        risiko = "tinggi"

# Meningkatkan risiko satu tingkat jika tidak memiliki sistem pengaman
if sistem_pengaman == "tidak":
    if risiko == "rendah":
        risiko = "sedang"
    elif risiko == "sedang":
        risiko = "tinggi"

# Menampilkan kategori risiko kendaraan
print("Kategori risiko kendaraan:", risiko)
```

```python
# Memasukkan nilai kendaraan
nilai_kendaraan = int(input("Masukkan nilai kendaraan (Rp): "))

# Memasukkan kategori risiko kendaraan
risiko = input("Masukkan kategori risiko (rendah/sedang/tinggi): ").lower()

# Menentukan persentase premi berdasarkan risiko
if risiko == "tinggi":
    persentase_premi = 0.1
elif risiko == "sedang":
    persentase_premi = 0.07
else:
    persentase_premi = 0.05

# Menghitung premi asuransi
premi_asuransi = persentase_premi * nilai_kendaraan

# Menampilkan premi asuransi
print("Premi asuransi yang perlu dibayarkan: Rp{:,.0f}".format(premi_asuransi))
```
