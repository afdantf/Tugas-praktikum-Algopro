```python
hari = 1
saldo = 50000

while hari <= 30:
    print("Hari ke-", hari)
    pengeluaran = int(input("Masukkan pengeluaran Nobita: Rp"))

    saldo -= pengeluaran

    if saldo < 0:
        print("Saldo Nobita habis!")
        break

    print("Sisa saldo Nobita: Rp", saldo)
    hari += 1

else:
    print("Pencatatan pengeluaran selama 30 hari selesai.")

print("Saldo akhir Nobita: Rp", saldo)
```

```python
total_keseluruhan = 0

for bulan in range(1, 4):
    total_bulanan = 0
    print("\nBulan ke-", bulan)

    for minggu in range(1, 5):
        pengeluaran = int(input(f"Pengeluaran minggu ke-{minggu}: Rp"))

        if pengeluaran > 20000:
            print("Pengeluaran melebihi Rp20.000. Pencatatan bulan dihentikan.")
            break

        total_bulanan += pengeluaran

    else:
        print("Pencatatan bulan selesai tanpa pelanggaran.")

    print("Total pengeluaran bulan ke-", bulan, ": Rp", total_bulanan)
    total_keseluruhan += total_bulanan

print("\n=== RINGKASAN KEUANGAN NOBITA ===")
print("Total pengeluaran selama 3 bulan: Rp", total_keseluruhan)
```

```python
saldo_awal = 300000
saldo = saldo_awal

for bulan in range(11, 13):
    print("\nBulan ke-", bulan)
    minggu = 1

    while minggu <= 4:
        libur = input(f"Apakah minggu ke-{minggu} libur (y/n)? ")

        if libur == "y":
            print("Minggu libur, pencatatan dilewati.")
            minggu += 1
            continue

        elif libur == "n":
            pengeluaran = int(input("Masukkan pengeluaran: Rp"))
            saldo -= pengeluaran

            if saldo < 0:
                print("Saldo Nobita negatif. Simulasi dihentikan!")
                break

            print("Sisa saldo Nobita: Rp", saldo)

        minggu += 1

    else:
        print("Saldo akhir bulan ke-", bulan, ": Rp", saldo)

    if saldo < 0:
        break

else:
    print(f"\nNobita berhasil menabung Rp{saldo} dari total Rp{saldo_awal}.")
```

```python
saldo_awal = 300000
saldo = saldo_awal

for bulan in range(11, 13):
    print("\nBulan ke-", bulan)
    minggu = 1

    while minggu <= 4:
        libur = input(f"Apakah minggu ke-{minggu} libur (y/n)? ")

        if libur.lower() == "y":
            print("Minggu libur, pencatatan dilewati.")
            minggu += 1
            continue

        elif libur.lower() == "n":
            pengeluaran = int(input("Masukkan pengeluaran: Rp"))
            saldo -= pengeluaran

            if saldo < 0:
                print("Saldo negatif! Simulasi dihentikan.")
                break

            print("Sisa saldo Nobita: Rp", saldo)

        else:
            print("Input tidak valid. Masukkan y atau n.")
            continue

        minggu += 1

    else:
        print(f"Saldo akhir bulan ke-{bulan}: Rp{saldo}")

    if saldo < 0:
        break

else:
    print(
        f"\nNobita berhasil menabung Rp{saldo} "
        f"dari total Rp{saldo_awal}."
    )
```
