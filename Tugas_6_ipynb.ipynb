```python
# a. Membuat variabel global
skor_nobita = 0


# b. Fungsi untuk memperbarui skor
def update_skor(nilai):
    global skor_nobita
    skor_nobita += nilai


# c.1 Fungsi belajar
def belajar(jam_belajar):
    if jam_belajar < 1:
        print("Nobita hanya membuka buku sebentar")
        skor = 50
    elif jam_belajar < 3:
        print("Nobita belajar secukupnya")
        skor = 25
    else:
        print("Nobita belajar dengan rajin hari ini")
        skor = 10

    return skor


# Contoh penggunaan fungsi
jam_belajar = float(input("Berapa jam Nobita belajar hari ini? "))

nilai = belajar(jam_belajar)
update_skor(nilai)

print("Skor belajar Nobita:", nilai)
print("Total skor Nobita:", skor_nobita)
```

```python
# Fungsi untuk menghitung skor tidur Nobita
def tidur(jam_tidur):
    if jam_tidur < 1:
        print("Nobita kurang tidur")
        skor = -10
    elif jam_tidur <= 2:
        print("Waktu tidur Nobita seimbang")
        skor = 20
    else:
        print("Nobita tidur terlalu lama")
        skor = -20

    return skor


# Contoh penggunaan fungsi
jam_tidur = float(input("Berapa jam Nobita tidur siang hari ini? "))

nilai = tidur(jam_tidur)
update_skor(nilai)

print("Skor tidur Nobita:", nilai)
print("Total skor Nobita:", skor_nobita)
```

```python
# Fungsi untuk menghitung skor bermain Nobita
def main_bersama_doraemon(jumlah_gadget, waktu_main):
    if waktu_main <= 1 and jumlah_gadget <= 2:
        print("Nobita bermain secukupnya")
        skor = 15
    elif waktu_main <= 2 and jumlah_gadget <= 3:
        print("Nobita masih bisa mengatur waktu dengan baik")
        skor = 5
    else:
        print("Nobita kebanyakan bermain")
        skor = -30

    return skor


# Contoh penggunaan fungsi
jumlah_gadget = int(input("Berapa gadget yang digunakan Nobita? "))
waktu_main = float(input("Berapa jam Nobita bermain gadget? "))

nilai = main_bersama_doraemon(jumlah_gadget, waktu_main)
update_skor(nilai)

print("Skor bermain Nobita:", nilai)
print("Total skor Nobita:", skor_nobita)
```

```python
# Fungsi untuk aktivitas menyontek
def menyontek():
    print("Nobita ketahuan menyontek")
    skor = -50
    return skor


# Contoh penggunaan fungsi
nilai = menyontek()
update_skor(nilai)

print("Skor menyontek Nobita:", nilai)
print("Total skor Nobita:", skor_nobita)
```

```python
# Fungsi untuk menjalankan aktivitas Nobita
def jalankan_aktivitas(pilihan):
    if pilihan == 1:
        jam_belajar = float(input("Berapa jam Nobita belajar hari ini? "))
        nilai = belajar(jam_belajar)

    elif pilihan == 2:
        jam_tidur = float(input("Berapa jam Nobita tidur siang hari ini? "))
        nilai = tidur(jam_tidur)

    elif pilihan == 3:
        jumlah_gadget = int(input("Berapa gadget yang digunakan Nobita? "))
        waktu_main = float(input("Berapa jam Nobita bermain gadget? "))
        nilai = main_bersama_doraemon(jumlah_gadget, waktu_main)

    elif pilihan == 4:
        nilai = menyontek()

    else:
        print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")
        return

    update_skor(nilai)
    print("Skor aktivitas:", nilai)
    print("Total skor Nobita:", skor_nobita)
```

```python
# Fungsi untuk mengevaluasi perilaku Nobita
def evaluasi_akhir(skor):
    if skor >= 80:
        return "Siswa Teladan"
    elif skor >= 50:
        return "Cukup Rajin"
    else:
        return "Pemalas"


# Contoh penggunaan fungsi
hasil = evaluasi_akhir(skor_nobita)

print("Hasil evaluasi Nobita:", hasil)
```

```python
# Fungsi menu utama
def menu_utama():
    while True:
        print("\n===== PROGRAM PENINGKAT DISIPLIN HARIAN =====")
        print("1. Belajar")
        print("2. Tidur")
        print("3. Bermain bersama Doraemon")
        print("4. Menyontek")
        print("5. Evaluasi Akhir")
        print("============================================")

        pilihan = int(input("Pilih aktivitas (1-5): "))

        if pilihan >= 1 and pilihan <= 4:
            jalankan_aktivitas(pilihan)

        elif pilihan == 5:
            print("\n===== EVALUASI AKHIR NOBITA =====")
            print("Total skor Nobita:", skor_nobita)
            print("Hasil evaluasi:", evaluasi_akhir(skor_nobita))
            print("Program selesai.")
            break

        else:
            print("Pilihan tidak valid, aktivitas hanya ada 4 pilihan")


# Memanggil menu utama
menu_utama()
```
