```python
import os

# Membuat folder saving_project jika belum ada
os.makedirs('saving_project', exist_ok=True)

# Menulis file interest.py
interest_code = """def compound_interest(principal, rate, years):
    # Rumus: A = P * (1 + r)^t
    return principal * ((1 + rate) ** years)

def simple_interest(principal, rate, years):
    # Rumus: A = P * r * t + P (Nilai akhir tabungan dengan bunga sederhana)
    return principal + (principal * rate * years)
"""

with open('saving_project/interest.py', 'w') as f:
    f.write(interest_code)

print("Folder 'saving_project' dan file 'interest.py' berhasil dibuat.")
```

```python
# Menulis file display.py
display_code = """def print_result(principal, rate, years, ci, si):
    print("===========================================")
    print("      KALKULATOR BUNGA TABUNGAN NOBITA     ")
    print("===========================================")
    print(f"Saldo Awal           : ¥{principal:,.2f}")
    print(f"Bunga Tahunan        : {rate * 100:.1f}%")
    print(f"Lama Menabung        : {years} tahun")
    print("-------------------------------------------")
    print(f"Nilai Akhir (Bunga Majemuk)   : ¥{ci:,.2f}")
    print(f"Nilai Akhir (Bunga Sederhana) : ¥{si:,.2f}")
    print("===========================================")
"""

with open('saving_project/display.py', 'w') as f:
    f.write(display_code)

print("File 'display.py' berhasil dibuat.")
```

```python
# Menulis file main.py
main_code = """import interest
import display

def main():
    # Parameter data tabungan Nobita
    principal = 80000
    rate = 0.05
    years = 4

    # Melakukan perhitungan bunga menggunakan modul interest
    ci_result = interest.compound_interest(principal, rate, years)
    si_result = interest.simple_interest(principal, rate, years)

    # Menampilkan hasil dengan modul display
    display.print_result(principal, rate, years, ci_result, si_result)

if __name__ == '__main__':
    main()
"""

with open('saving_project/main.py', 'w') as f:
    f.write(main_code)

print("File 'main.py' berhasil dibuat.")
```
