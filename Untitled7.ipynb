```python
# Menyimpan nama pembeli ke dalam variabel
nama_pembeli = "Nobita Nobi"

# Menyimpan nama toko ke dalam variabel
nama_toko = "Toko Sembako Doraemon"

# Membuat list yang berisi nama barang belanjaan
belanjaan = ["Apel", "Susu", "Roti"]

# Membuat tuple yang berisi harga dan jumlah setiap barang
harga_jumlah = (5000, 3, 12000, 2, 8000, 1)

# Menghitung total harga setiap barang
total_apel = harga_jumlah[0] * harga_jumlah[1]
total_susu = harga_jumlah[2] * harga_jumlah[3]
total_roti = harga_jumlah[4] * harga_jumlah[5]

# Menghitung total seluruh belanjaan
total_belanja = total_apel + total_susu + total_roti

# Mencetak struk menggunakan f-string dan escape sequences
print("=" * 40)
print(f"\t{nama_toko}")
print("=" * 40)
print(f"Nama Pembeli : {nama_pembeli}")
print("-" * 40)
print("Nama Barang\tHarga\tJumlah\tTotal")
print("-" * 40)
print(f"{belanjaan[0]}\t\t{harga_jumlah[0]}\t{harga_jumlah[1]}\t{total_apel}")
print(f"{belanjaan[1]}\t\t{harga_jumlah[2]}\t{harga_jumlah[3]}\t{total_susu}")
print(f"{belanjaan[2]}\t\t{harga_jumlah[4]}\t{harga_jumlah[5]}\t{total_roti}")
print("-" * 40)
print(f"Total Belanja: Rp{total_belanja:,}".replace(",", "."))
print("=" * 40)
print("\tTerima kasih sudah berbelanja!")
print("=" * 40)
```
