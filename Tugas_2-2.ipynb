```
berat_badan = 97.5
tinggi_badan = 1.75
bmi = berat_badan / (tinggi_badan ** 2)
print("bmi asli:", bmi)
bmi_bulat = round(bmi, 2)
print("bmi dibulatkan:", bmi_bulat)
```

```
target_air = 2000
kapasitas_botol = 350
botol_desimal = target_air / kapasitas_botol
botol_penuh = target_air // kapasitas_botol
sisa_air = target_air % kapasitas_botol

print("hasil true division:", botol_desimal)
print("botol penuh:", botol_penuh)
print("sisa air:", sisa_air, "ml")
```

```
import math

kalori_harian = 2100.5
kalori_bawah = math.floor(kalori_harian)
kalori_atas = math.ceil(kalori_harian)

print("kalori dibulatkan kebawah:", kalori_bawah)
print("kalori dibulatkan keatas:", kalori_atas)
```
